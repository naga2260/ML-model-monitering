```python
"""
House price dataset — data cleaning pipeline.

Goal: fix data-entry errors and tame skew/outliers WITHOUT deleting rows.
Run: df_clean = clean_pipeline(df)
"""

import numpy as np
import pandas as pd

# ---------------------------------------------------------------
# Column groups — outlier treatment must match the column's nature.
# Applying one rule to every numeric column is the most common
# cleaning mistake (e.g. IQR on a binary/ordinal column is meaningless).
# ---------------------------------------------------------------
AREA_FEATURES = [
    "living area", "lot area", "Area of the house(excluding basement)",
    "living_area_renov", "lot_area_renov",
]
COUNT_FEATURES = ["number of bedrooms", "number of bathrooms"]
ZERO_INFLATED = ["Area of the basement", "Renovation Year"]
BOUNDED_DISCRETE = [  # already a fixed, finite scale — never outlier-cap these
    "number of floors", "waterfront present", "number of views",
    "condition of the house", "grade of the house", "Number of schools nearby",
]


def audit(df):
    """Look before you clean — confirms what actually needs fixing."""
    print("shape:", df.shape)
    miss = df.isnull().sum()
    print("missing values:", miss[miss > 0].to_dict() or "none")
    print("duplicate rows:", df.duplicated().sum())
    return df


def drop_exact_duplicates(df):
    """Removing an identical copy of a row isn't 'losing data' — it's
    redundant information. This is the one place row deletion is fine."""
    before = len(df)
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"removed {before - len(df)} exact duplicate rows")
    return df


def fix_logical_inconsistencies(df):
    """A renovation year earlier than the build year is impossible — it's a
    data-entry error, not a statistical outlier. Correct it (treat as
    'not renovated') instead of deleting the row and everything else it knows."""
    df = df.copy()
    bad = (df["Renovation Year"] > 0) & (df["Renovation Year"] < df["Built Year"])
    print(f"renovation-before-built rows fixed: {bad.sum()}")
    df.loc[bad, "Renovation Year"] = 0
    return df


def flag_bedroom_inconsistencies(df, min_sqft_per_bedroom=150):
    """Judge 'number of bedrooms' against living area, not in isolation.
    A high bedroom count is only a real problem if the square footage
    can't plausibly support it (e.g. 33 bedrooms in a 2,000 sqft home)."""
    ratio = df["living area"] / df["number of bedrooms"].replace(0, np.nan)
    suspicious = ratio < min_sqft_per_bedroom
    print(f"bedroom count inconsistent with living area: {suspicious.sum()} rows")
    return suspicious


def encode_zero_inflated(df):
    """0 in basement area / renovation year means 'not applicable', not a
    low value on a continuous scale. Split into a flag + the original value
    so the zero stops getting treated as a statistical outlier downstream."""
    df = df.copy()
    df["has_basement"] = (df["Area of the basement"] > 0).astype(int)
    df["is_renovated"] = (df["Renovation Year"] > 0).astype(int)
    return df


def winsorize(df, columns, lower_pct=0.5, upper_pct=99.5):
    """Clip each column to its [lower_pct, upper_pct] percentile range.
    Every row is kept — only the extreme value itself is pulled in.
    This is the direct 'don't lose data' alternative to dropping rows."""
    df = df.copy()
    for col in columns:
        lo, hi = np.percentile(df[col], [lower_pct, upper_pct])
        n_affected = int(((df[col] < lo) | (df[col] > hi)).sum())
        df[col] = df[col].clip(lo, hi)
        print(f"{col}: clipped {n_affected} rows to [{lo:.1f}, {hi:.1f}]")
    return df


def add_log_features(df, columns):
    """Right-skewed area features (lot area's max is ~70x its 75th
    percentile) get a log1p companion column, which tames the skew for
    linear regression without discarding or overwriting the raw values."""
    df = df.copy()
    for col in columns:
        df[f"{col}_log"] = np.log1p(df[col])
    return df


def clean_pipeline(df):
    df = audit(df)
    df = drop_exact_duplicates(df)
    df = fix_logical_inconsistencies(df)

    suspicious = flag_bedroom_inconsistencies(df)
    if suspicious.any():
        print(df.loc[suspicious, ["number of bedrooms", "living area"]])

    df = encode_zero_inflated(df)

    # count-like columns: conservative percentile bounds so we only
    # touch genuinely extreme entries, not legitimate large homes
    df = winsorize(df, COUNT_FEATURES, lower_pct=0.1, upper_pct=99.9)

    # skewed area columns: slightly wider percentile bounds
    df = winsorize(df, AREA_FEATURES, lower_pct=0.5, upper_pct=99.5)

    df = add_log_features(df, ["lot area", "lot_area_renov"])

    print("\nfinal shape:", df.shape)
    return df


# ---------------------------------------------------------------
# Usage (in your notebook, after you've already loaded df):
#
#   df_clean = clean_pipeline(df)
#
# Note: BOUNDED_DISCRETE columns (floors, waterfront, views, condition,
# grade, schools nearby) and the target (Price) are intentionally left
# untouched here — see the explanation for why.
# ---------------------------------------------------------------
```

```python
"""
Feature engineering for the house price dataset.
Run after clean_house_data.clean_pipeline():

    df_clean = clean_pipeline(df)
    df_fe = engineer_features(df_clean)
"""

import numpy as np
import pandas as pd
from sklearn.cluster import KMeans

REFERENCE_YEAR = 2015  # dataset's built-year ceiling; swap for the real sale year if you have one


def add_age_features(df, reference_year=REFERENCE_YEAR):
    """
    Built Year -> age is an AFFINE transform (age = ref - built_year), so on
    its own it will NOT change a linear model's R2 or RMSE -- same
    information, flipped sign. It's worth doing for interpretability, but
    the real gain is 'effective_age': age since the MORE RECENT of build or
    renovation. A linear model can't derive that from the raw year columns
    on its own, because max() is a nonlinear operation.
    """
    df = df.copy()
    df["house_age"] = reference_year - df["Built Year"]
    effective_year = np.where(df["is_renovated"] == 1, df["Renovation Year"], df["Built Year"])
    df["effective_age"] = reference_year - effective_year
    return df


def check_redundant_areas(df):
    """
    living area ~= Area of the house(excluding basement) + Area of the
    basement -- their means matched almost to the decimal in your
    describe() output. If the row-by-row difference below is ~0, you have a
    hard collinearity: one column is fully determined by the other two.
    That destabilizes coefficients even though it won't hurt R2 outright.
    """
    diff = df["Area of the house(excluding basement)"] + df["Area of the basement"] - df["living area"]
    print("living area vs (house area + basement area), difference stats:")
    print(diff.describe())
    return diff


def check_renov_columns(df):
    """
    Don't assume from the column name what *_renov actually encodes.
    Compare it against the base column for NON-renovated rows: if they
    match there, the renov columns are likely just an updated value for
    renovated homes (safe to fold together). If they differ even when
    is_renovated == 0, they're measuring something else (e.g. a
    neighborhood-comparable size) and are worth keeping as separate signal.
    """
    subset = df.loc[df["is_renovated"] == 0, ["living area", "living_area_renov"]]
    print("living area vs living_area_renov, non-renovated rows only:")
    print((subset["living area"] - subset["living_area_renov"]).describe())


def add_location_features(df, n_clusters=15, random_state=42):
    """
    Raw lat/long in a LINEAR model only gives a straight-line price
    gradient (price rising steadily north or east) -- it structurally
    cannot represent 'this pocket of the map is a pricier neighborhood',
    which is how real estate pricing actually behaves. Two fixes:

      1. distance_from_center: one feature capturing 'closer to the core
         costs more' -- a real, common pattern.
      2. location_cluster: KMeans groups nearby homes; one-hot encoding the
         cluster lets the model learn a distinct baseline price per area,
         something linear lat/long terms alone cannot do.
    """
    df = df.copy()
    center_lat, center_lon = df["Lattitude"].mean(), df["Longitude"].mean()
    df["distance_from_center"] = np.sqrt(
        (df["Lattitude"] - center_lat) ** 2 + (df["Longitude"] - center_lon) ** 2
    )

    km = KMeans(n_clusters=n_clusters, random_state=random_state, n_init=10)
    df["location_cluster"] = km.fit_predict(df[["Lattitude", "Longitude"]])
    dummies = pd.get_dummies(df["location_cluster"], prefix="loc_cluster", drop_first=True)
    df = pd.concat([df, dummies], axis=1)
    return df


def add_log_target(df, price_col="Price"):
    """
    Price is heavily right-skewed (max ~14x the mean). Linear regression
    assumes roughly-additive, roughly-normal residuals, so a handful of very
    expensive homes otherwise dominate the error. Training on log1p(Price)
    turns multiplicative effects ('an extra bathroom adds X%') into additive
    ones, which is what the model actually assumes. Remember to
    np.expm1() your predictions back to dollars before computing MAE/RMSE,
    so the error metrics stay interpretable.
    """
    df = df.copy()
    df["log_price"] = np.log1p(df[price_col])
    return df


def engineer_features(df):
    df = add_age_features(df)
    check_redundant_areas(df)     # inspect the printed output before dropping anything
    check_renov_columns(df)       # same -- decide based on what this shows, don't assume
    df = add_location_features(df)
    df = add_log_target(df)

    # Drop the columns now made redundant by the engineered features above.
    # (Area of the house / Area of the basement are kept -- they're more
    # granular than the sum, and let the model weight main-floor vs
    # basement sqft differently.)
    df = df.drop(columns=["living area", "Built Year", "Renovation Year"])

    return df


# ---------------------------------------------------------------
# Usage:
#   df_clean = clean_pipeline(df)          # from clean_house_data.py
#   df_fe = engineer_features(df_clean)
#
#   X = df_fe.drop(columns=['Price', 'log_price'])
#   y = df_fe['log_price']                 # train on log_price
#   ...
#   preds_dollars = np.expm1(model.predict(X_test))   # back-transform for MAE/RMSE
# ---------------------------------------------------------------
```
