```python
import pandas as pd

# Make a copy of your DataFrame to avoid overwriting original data
df_clean = df.copy()

# List of continuous or numeric columns where you want to remove extreme outliers
# (Exclude binary columns like 'waterfront present' or 'Has_Basement')
columns_to_check = [
    'number of bedrooms', 
    'number of bathrooms', 
    'living area', 
    'lot area', 
    'Area of the basement'
]

# Function to remove outliers using IQR
for col in columns_to_check:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Define lower and upper bounds (using 1.5 as the standard multiplier)
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter the dataframe to keep only rows within the bounds for this column
    df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]

print(f"Original shape: {df.shape}")
print(f"Shape after removing IQR outliers: {df_clean.shape}")
```
